## 2. Reading the Drawn Points

After drawing, read the saved curve points:

```python
import json
import numpy as np

# Read the points saved by the drawing interface
with open("curve_points.json", "r") as file:
    points = json.load(file)

# Get the last drawn curve segment
if points:
    latest_curve = np.array(points[-1])
    print(f"Drew {len(latest_curve)} points")
    print(f"First point: {latest_curve[0] if len(latest_curve) > 0 else None}")
    print(f"Last point: {latest_curve[-1] if len(latest_curve) > 0 else None}")
```

## 3. Using the Shared NURBS Functions

Instead of duplicating the NURBS function, use the shared geological module:

```python
# Replace duplicated NURBS function with:
from shared.geological import create_nurbs_curve

# Example: Create a salt dome geometry using NURBS
curve_points, derivative_angles = create_nurbs_curve(
    left_x=0, right_x=20, mid_left_x=3, mid_right_x=17,
    top_left_x=7, top_right_x=13, top_x=10,
    left_y=5, right_y=5, mid_left_y=5, mid_right_y=5,
    top_left_y=5, top_right_y=5, top_y=5,
    nr_points=50, fig=True
)

print(f"Generated {len(curve_points)} curve points")
```

## 4. Using the Shared Salt Formation Class

Instead of duplicating the salt class, use the shared formation module:

```python
# Replace duplicated salt class with:
from shared.geological import SaltFormation

# Create two slices for the salt formation
slice_1_parameter = (curve_points, derivative_angles)
slice_2_parameter = (curve_points, derivative_angles)  # In practice, use different curves

# Create salt formation
salt_formation = SaltFormation(
    slice_1_parameter=slice_1_parameter,
    slice_2_parameter=slice_2_parameter,
    slice_1_pos=10,
    slice_2_pos=10,
    formation_name="salt_dome"
)

# Create and compute the geological model
geo_model = salt_formation.compute_model(plot_3d=True)
```
